```python
from pyspark.sql import functions as F
```

##==========================================
###READING CSV/JSON FILES FROM VOLUMES
##==========================================

```python

input_file = '/Volumes/first_spark_catalog/bronze_first_spark/first_spark_volume/employees_1gb.csv'

df = (
     spark.read
     .option("header",True)
     .option("inferSchema",True)
     .csv(input_file)
)

df.printSchema()

df.show(10,truncate=False)
```

###===================================================
##BASIC DATA QUALITY AND DISPLAING TOTAL RECORDS COUNT
###===================================================

```python
print('Total No of Records',df.count())
df.display(20,truncate=False)
```

###### ============================================================
### 4. CLEAN STRING COLUMNS
###### ============================================================

```python
df_clean = (
    df
    .withColumn("first_name",F.trim(F.col("first_name")))
    .withColumn("last_name",F.trim(F.col("last_name")))
    .withColumn("region",F.trim(F.col("region")))
)

df_clean.show(20,truncate=False) 
```

######============================================================
### 5. DROPING / REMOVING DUPLICATES
######============================================================

```python
df_clean_ND = (
    df_clean
    .dropDuplicates(["emp_id"])    
)

df_clean_ND.show(20,truncate=False)

```

######============================================================
### 6. HANDLE NULLS
######============================================================

```python
df_clean_RM_NULLS = (
    df_clean_ND
    .fillna({"first_name":"Unknown",
             "last_name":"Unknown",
             "department":"Unknown",
             "region":"Unknown",
             "salary":0
             })
)

df_clean_RM_NULLS.show(10,truncate=False)
```
